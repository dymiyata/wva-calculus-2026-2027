```python
import math
from math import atan, pi, sin, cos, atanh, cosh, e, sinh, log

m = 100

alphas = [atan(2 ** (-i)) for i in range(m)]

K = 1
for i in range(m):
    K *= 1 / (1 + 2 ** (-2 * i)) ** 0.5
```

```python
print(K)
```

```python
def cordic(theta): # return cos theta and sin theta

    # Reduce to between -pi and pi
    theta = (theta + pi) % (2 * pi) - pi

    # Get between -pi/2 and pi/2
    final_pm = 1
    if theta > pi / 2:
        theta -= pi
        final_pm = -1
    if theta < -pi / 2:
        theta += pi
        final_pm = -1


    x = 1
    y = 0
    phi = 0

    for i in range(m):

        # Do we add or subtract alpha?
        if phi < theta:
            pm = 1
        else:
            pm = -1

        phi += pm * alphas[i]

        new_x = x - pm * y * 2 ** (-i)
        new_y = y + pm * x * 2 ** (-i)

        x = new_x
        y = new_y

    return final_pm * K * x , final_pm * K * y
```

```python
m = 200

repeated = [4, 13, 40, 121]

indices = []
for i in range(1, m):
    indices.append(i)
    if i in repeated:
        indices.append(i)

alphas_h = [atanh(2 ** (-i)) for i in range(1, m)]

K_h = 1
for i in indices:
    K_h *= 1 / (1 - 2 ** (-2 * i)) ** 0.5 # cosh of tanh^-1(2^-i)
```

```python
print(indices)
```

```python
K_h
```

```python
def cordic_h(theta):
    x = 1
    y = 0
    phi = 0

    for i in indices:

        # Do we add or subtract alpha?
        if phi < theta:
            pm = 1
        else:
            pm = -1

        phi += pm * alphas_h[i - 1]

        new_x = x + pm * y * 2 ** (-i)
        new_y = y + pm * x * 2 ** (-i)

        x = new_x
        y = new_y

    return  K_h * x ,  K_h * y
```

```python
cordic_h(0.5)
```

```python
def natural_exp(x):
    q = math.floor(x / log(2))
    r = x - q * log(2)

    a, b = cordic_h(r)
    er = a + b

    return 2 ** q * er
```
